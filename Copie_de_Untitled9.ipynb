```python
import pandas as pd
import numpy as np
import zipfile
from pathlib import Path

ZIP_PATH = "/content/drive/MyDrive/Competition/Stage_II_B_31_r1_f1.zip"

with zipfile.ZipFile(ZIP_PATH) as z:

    members = z.namelist()

    def find_file(filename):
        matches = [m for m in members if m.endswith(filename)]
        if len(matches) != 1:
            raise FileNotFoundError(
                f"Expected one {filename}; found {matches}"
            )
        return matches[0]

    train = pd.read_csv(
        z.open(find_file("Dependency_Train_OOF.csv.gz")),
        compression="gzip"
    )

    valid = pd.read_csv(
        z.open(find_file("Dependency_Validation.csv.gz")),
        compression="gzip"
    )

    metrics = pd.read_csv(
        z.open(find_file("Dependency_Metrics.csv"))
    )

    feature_file = find_file("Feature_Column_List.txt")
    feature_cols = (
        z.read(feature_file).decode("utf-8").splitlines()
    )

print("TRAIN:", train.shape)
print("VALIDATION:", valid.shape)
print("GENERATED FEATURES:", len(feature_cols))

# --------------------------------------------------
# Integrity checks
# --------------------------------------------------

checks = {
    "Expected training rows": len(train) == 136961,
    "Expected validation rows": len(valid) == 34241,
    "Expected generated features": len(feature_cols) == 48,
    "Same train/validation columns": list(train.columns) == list(valid.columns),
    "No TARGET column": "TARGET" not in train.columns and "TARGET" not in valid.columns,
    "Unique training row indices": train["row_index"].is_unique,
    "Unique validation row indices": valid["row_index"].is_unique,
    "Unique training applicant IDs": train["SK_ID_CURR"].is_unique,
    "Unique validation applicant IDs": valid["SK_ID_CURR"].is_unique,
}

print("\nINTEGRITY CHECKS")
for name, passed in checks.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

# --------------------------------------------------
# Feature distribution diagnostics
# --------------------------------------------------

def psi_10bin(reference, current):
    """Decile-bin PSI, with missing values as a separate bin."""
    a = np.asarray(reference, dtype=float)
    b = np.asarray(current, dtype=float)

    a = a[np.isfinite(a)]
    b = b[np.isfinite(b)]

    if len(a) == 0 or len(b) == 0:
        return np.nan

    internal = np.unique(
        np.quantile(a, np.linspace(0, 1, 11))
    )[1:-1]

    ca = np.bincount(
        np.searchsorted(internal, a, side="right"),
        minlength=len(internal) + 1
    ).astype(float)

    cb = np.bincount(
        np.searchsorted(internal, b, side="right"),
        minlength=len(internal) + 1
    ).astype(float)

    # Add one missing-value bin.
    ca = np.append(ca, np.sum(~np.isfinite(reference)))
    cb = np.append(cb, np.sum(~np.isfinite(current)))

    pa = (ca + 1e-6) / (ca.sum() + 1e-6 * len(ca))
    pb = (cb + 1e-6) / (cb.sum() + 1e-6 * len(cb))

    return float(np.sum((pa - pb) * np.log(pa / pb)))


rows = []

for col in feature_cols:

    a = pd.to_numeric(train[col], errors="coerce").to_numpy(float)
    b = pd.to_numeric(valid[col], errors="coerce").to_numpy(float)

    af = a[np.isfinite(a)]
    bf = b[np.isfinite(b)]

    q = np.quantile(af, [0.01, 0.50, 0.99]) if len(af) else [np.nan]*3

    rows.append({
        "feature": col,
        "train_missing_pct": 100 * np.mean(~np.isfinite(a)),
        "valid_missing_pct": 100 * np.mean(~np.isfinite(b)),
        "train_mean": np.mean(af) if len(af) else np.nan,
        "valid_mean": np.mean(bf) if len(bf) else np.nan,
        "train_std": np.std(af) if len(af) else np.nan,
        "valid_std": np.std(bf) if len(bf) else np.nan,
        "train_median": np.median(af) if len(af) else np.nan,
        "valid_median": np.median(bf) if len(bf) else np.nan,
        "train_p01": q[0],
        "train_p99": q[2],
        "valid_below_train_p01_pct":
            100 * np.mean(bf < q[0]) if len(bf) else np.nan,
        "valid_above_train_p99_pct":
            100 * np.mean(bf > q[2]) if len(bf) else np.nan,
        "train_zero_pct": 100 * np.mean(af == 0) if len(af) else np.nan,
        "valid_zero_pct": 100 * np.mean(bf == 0) if len(bf) else np.nan,
        "train_unique": len(np.unique(af)),
        "valid_unique": len(np.unique(bf)),
        "PSI_10bin": psi_10bin(a, b),
    })

summary = pd.DataFrame(rows).sort_values(
    "PSI_10bin", ascending=False
)

# Save compact reports in /content.
OUT = Path("/content/Stage_II_B_Compact_Diagnostic")
OUT.mkdir(exist_ok=True)

summary.to_csv(OUT / "Feature_Distribution_Summary.csv", index=False)
metrics.to_csv(OUT / "Dependency_Metrics.csv", index=False)

(OUT / "Integrity_Checks.txt").write_text(
    "\n".join(f"{k}: {'PASS' if v else 'FAIL'}"
              for k, v in checks.items()),
    encoding="utf-8"
)

print("\nDEPENDENCY METRICS")
print(metrics.to_string(index=False))

print("\nFEATURE DISTRIBUTIONS — HIGHEST PSI FIRST")
print(summary.round(5).to_string(index=False))

print("\nReports saved to:", OUT)
```

```python
import pandas as pd
import numpy as np
import zipfile
from pathlib import Path

# ============================================================
# CONFIGURATION
# ============================================================

ZIP_DIR = Path("/content/drive/MyDrive/Competition")

ZIP_PATHS = [
    ZIP_DIR / "Stage_II_B_31_r1_f1.zip",
    ZIP_DIR / "Stage_II_B_31_r1_f2.zip",
    ZIP_DIR / "Stage_II_B_31_r1_f3.zip",
    ZIP_DIR / "Stage_II_B_31_r1_f4.zip",
    ZIP_DIR / "Stage_II_B_31_r1_f5.zip",
]

OUT = Path("/content/Stage_II_B_Compact_Diagnostic_r1")
OUT.mkdir(exist_ok=True)


# ============================================================
# ZIP READER
# ============================================================

def find_file(members, filename):
    matches = [m for m in members if m.endswith(filename)]

    if len(matches) != 1:
        raise FileNotFoundError(
            f"Expected one {filename}; found {matches}"
        )

    return matches[0]


def read_fold(zip_path):

    print("\n" + "=" * 80)
    print("READING:", zip_path.name)
    print("=" * 80)

    with zipfile.ZipFile(zip_path) as z:

        members = z.namelist()

        train = pd.read_csv(
            z.open(find_file(
                members,
                "Dependency_Train_OOF.csv.gz"
            )),
            compression="gzip"
        )

        valid = pd.read_csv(
            z.open(find_file(
                members,
                "Dependency_Validation.csv.gz"
            )),
            compression="gzip"
        )

        metrics = pd.read_csv(
            z.open(find_file(
                members,
                "Dependency_Metrics.csv"
            ))
        )

        feature_file = find_file(
            members,
            "Feature_Column_List.txt"
        )

        feature_cols = (
            z.read(feature_file)
            .decode("utf-8")
            .splitlines()
        )

    print("TRAIN:", train.shape)
    print("VALID:", valid.shape)
    print("GENERATED FEATURES:", len(feature_cols))

    return train, valid, metrics, feature_cols


# ============================================================
# PSI
# ============================================================

def psi_10bin(reference, current):

    a = np.asarray(reference, dtype=float)
    b = np.asarray(current, dtype=float)

    a_finite = np.isfinite(a)
    b_finite = np.isfinite(b)

    af = a[a_finite]
    bf = b[b_finite]

    if len(af) == 0 or len(bf) == 0:
        return np.nan

    # Train quantile boundaries
    internal = np.unique(
        np.quantile(
            af,
            np.linspace(0, 1, 11)
        )
    )[1:-1]

    ca = np.bincount(
        np.searchsorted(
            internal,
            af,
            side="right"
        ),
        minlength=len(internal) + 1
    ).astype(float)

    cb = np.bincount(
        np.searchsorted(
            internal,
            bf,
            side="right"
        ),
        minlength=len(internal) + 1
    ).astype(float)

    # Missing-value bin
    ca = np.append(
        ca,
        np.sum(~a_finite)
    )

    cb = np.append(
        cb,
        np.sum(~b_finite)
    )

    eps = 1e-6

    pa = (ca + eps) / (
        ca.sum() + eps * len(ca)
    )

    pb = (cb + eps) / (
        cb.sum() + eps * len(cb)
    )

    return float(
        np.sum(
            (pa - pb) *
            np.log(pa / pb)
        )
    )


# ============================================================
# STORAGE
# ============================================================

all_metrics = []
all_distributions = []
all_integrity = []


# ============================================================
# PROCESS FIVE FOLDS
# ============================================================

for fold_number, zip_path in enumerate(
    ZIP_PATHS,
    start=1
):

    train, valid, metrics, feature_cols = read_fold(
        zip_path
    )

    fold_name = f"r1_f{fold_number}"

    # --------------------------------------------------------
    # Integrity checks
    # --------------------------------------------------------

    checks = {
        "Expected training rows":
            len(train) in [136960, 136961],

        "Expected validation rows":
            len(valid) in [34240, 34241],

        "Expected generated features":
            len(feature_cols) == 48,

        "Same train/validation columns":
            list(train.columns) == list(valid.columns),

        "No TARGET":
            "TARGET" not in train.columns
            and "TARGET" not in valid.columns,

        "Unique training row indices":
            train["row_index"].is_unique,

        "Unique validation row indices":
            valid["row_index"].is_unique,

        "Unique training applicant IDs":
            train["SK_ID_CURR"].is_unique,

        "Unique validation applicant IDs":
            valid["SK_ID_CURR"].is_unique,

        "Train/valid generated feature count":
            len([
                c for c in train.columns
                if c not in [
                    "row_index",
                    "SK_ID_CURR"
                ]
            ]) == 48,
    }

    for name, passed in checks.items():

        all_integrity.append({
            "fold": fold_name,
            "check": name,
            "passed": passed
        })

    print("\nINTEGRITY:", fold_name)

    for name, passed in checks.items():

        print(
            f"{name}: "
            f"{'PASS' if passed else 'FAIL'}"
        )

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    metrics = metrics.copy()
    metrics.insert(0, "fold", fold_name)

    all_metrics.append(metrics)

    # --------------------------------------------------------
    # Feature distributions
    # --------------------------------------------------------

    for col in feature_cols:

        a = pd.to_numeric(
            train[col],
            errors="coerce"
        ).to_numpy(float)

        b = pd.to_numeric(
            valid[col],
            errors="coerce"
        ).to_numpy(float)

        af = a[np.isfinite(a)]
        bf = b[np.isfinite(b)]

        if len(af):

            train_p01 = np.quantile(
                af,
                0.01
            )

            train_p99 = np.quantile(
                af,
                0.99
            )

        else:

            train_p01 = np.nan
            train_p99 = np.nan

        all_distributions.append({

            "fold": fold_name,
            "feature": col,

            "train_missing_pct":
                100 * np.mean(~np.isfinite(a)),

            "valid_missing_pct":
                100 * np.mean(~np.isfinite(b)),

            "train_mean":
                np.mean(af)
                if len(af) else np.nan,

            "valid_mean":
                np.mean(bf)
                if len(bf) else np.nan,

            "train_std":
                np.std(af)
                if len(af) else np.nan,

            "valid_std":
                np.std(bf)
                if len(bf) else np.nan,

            "train_median":
                np.median(af)
                if len(af) else np.nan,

            "valid_median":
                np.median(bf)
                if len(bf) else np.nan,

            "train_p01":
                train_p01,

            "train_p99":
                train_p99,

            "valid_below_train_p01_pct":
                100 * np.mean(bf < train_p01)
                if len(bf) else np.nan,

            "valid_above_train_p99_pct":
                100 * np.mean(bf > train_p99)
                if len(bf) else np.nan,

            "train_zero_pct":
                100 * np.mean(af == 0)
                if len(af) else np.nan,

            "valid_zero_pct":
                100 * np.mean(bf == 0)
                if len(bf) else np.nan,

            "train_unique":
                len(np.unique(af)),

            "valid_unique":
                len(np.unique(bf)),

            "PSI_10bin":
                psi_10bin(a, b),
        })


# ============================================================
# COMBINE RESULTS
# ============================================================

metrics_all = pd.concat(
    all_metrics,
    ignore_index=True
)

dist_all = pd.DataFrame(
    all_distributions
)

integrity_all = pd.DataFrame(
    all_integrity
)


# ============================================================
# SAVE RAW COMBINED REPORTS
# ============================================================

metrics_all.to_csv(
    OUT / "All_Folds_Dependency_Metrics.csv",
    index=False
)

dist_all.to_csv(
    OUT / "All_Folds_Feature_Distributions.csv",
    index=False
)

integrity_all.to_csv(
    OUT / "All_Folds_Integrity_Checks.csv",
    index=False
)


# ============================================================
# METRIC SUMMARY
# ============================================================

# Automatically identify numeric metric columns
metric_numeric = metrics_all.select_dtypes(
    include=np.number
).columns.tolist()

metric_summary = (
    metrics_all
    .groupby(
        "metric",
        dropna=False
    )
    .agg(
        mean=("value", "mean"),
        std=("value", "std"),
        minimum=("value", "min"),
        maximum=("value", "max"),
    )
    .reset_index()
    if "metric" in metrics_all.columns
    else pd.DataFrame()
)

if len(metric_summary):

    metric_summary.to_csv(
        OUT / "Metric_CrossFold_Summary.csv",
        index=False
    )


# ============================================================
# FEATURE-LEVEL CROSS-FOLD SUMMARY
# ============================================================

feature_summary = (
    dist_all
    .groupby("feature")
    .agg(

        PSI_mean=(
            "PSI_10bin",
            "mean"
        ),

        PSI_std=(
            "PSI_10bin",
            "std"
        ),

        PSI_max=(
            "PSI_10bin",
            "max"
        ),

        train_missing_mean=(
            "train_missing_pct",
            "mean"
        ),

        valid_missing_mean=(
            "valid_missing_pct",
            "mean"
        ),

        missing_shift_mean=(
            "valid_missing_pct",
            lambda x:
            np.mean(
                x.values
                - dist_all.loc[
                    x.index,
                    "train_missing_pct"
                ].values
            )
        ),

        train_mean_mean=(
            "train_mean",
            "mean"
        ),

        valid_mean_mean=(
            "valid_mean",
            "mean"
        ),

        train_std_mean=(
            "train_std",
            "mean"
        ),

        valid_std_mean=(
            "valid_std",
            "mean"
        ),

        train_unique_min=(
            "train_unique",
            "min"
        ),

        valid_unique_min=(
            "valid_unique",
            "min"
        ),

    )
    .reset_index()
    .sort_values(
        "PSI_mean",
        ascending=False
    )
)

feature_summary.to_csv(
    OUT / "Feature_CrossFold_Summary.csv",
    index=False
)


# ============================================================
# FOLD CONSISTENCY OF PSI
# ============================================================

psi_pivot = dist_all.pivot(
    index="feature",
    columns="fold",
    values="PSI_10bin"
)

psi_pivot["PSI_mean"] = psi_pivot.mean(axis=1)
psi_pivot["PSI_std"] = psi_pivot.std(axis=1)
psi_pivot["PSI_max"] = psi_pivot.max(axis=1)

psi_pivot = psi_pivot.sort_values(
    "PSI_mean",
    ascending=False
)

psi_pivot.to_csv(
    OUT / "PSI_CrossFold_Matrix.csv"
)


# ============================================================
# IMPORTANT: TOP PSI FEATURES
# ============================================================

print("\n")
print("=" * 80)
print("TOP FEATURES BY MEAN PSI ACROSS REPETITION 1")
print("=" * 80)

print(
    feature_summary[
        [
            "feature",
            "PSI_mean",
            "PSI_std",
            "PSI_max",
            "train_missing_mean",
            "valid_missing_mean",
        ]
    ]
    .head(20)
    .round(5)
    .to_string(index=False)
)


# ============================================================
# IMPORTANT: METRICS
# ============================================================

print("\n")
print("=" * 80)
print("DEPENDENCY METRICS — ALL FIVE FOLDS")
print("=" * 80)

print(
    metrics_all.to_string(
        index=False
    )
)


# ============================================================
# INTEGRITY SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("INTEGRITY SUMMARY")
print("=" * 80)

integrity_summary = (
    integrity_all
    .groupby("check")["passed"]
    .agg(
        all_passed="all",
        failures=lambda x: int((~x).sum())
    )
    .reset_index()
)

print(
    integrity_summary.to_string(
        index=False
    )
)


# ============================================================
# FOLD COMPLETION SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("FOLD SUMMARY")
print("=" * 80)

for fold in sorted(
    dist_all["fold"].unique()
):

    d = dist_all[
        dist_all["fold"] == fold
    ]

    print(
        f"{fold}: "
        f"{len(d)} generated-feature diagnostics"
    )


print("\nReports saved to:")
print(OUT)
```

```python

import json
import zipfile
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.model_selection import RepeatedStratifiedKFold

# ------------------------------------------------------------
# Fixed values taken from the original baseline script
# ------------------------------------------------------------
BASE_SEED = 20261031
N_SPLITS = 5
N_REPEATS = 3

TARGET_COL = "TARGET"
ID_COL = "SK_ID_CURR"
EXCLUDE_COLUMNS = [ID_COL, "FLAG_MOBIL"]

TRAIN_PATH = Path(
    "/content/drive/MyDrive/Competition/input/train.csv"
)

ZIP_DIR = Path("/content/drive/MyDrive/Competition")

DEST = Path("/content/baseline_artifacts")
DEST.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Load the original dataset
# ------------------------------------------------------------
if not TRAIN_PATH.is_file():
    raise FileNotFoundError(f"Missing input: {TRAIN_PATH}")

df = pd.read_csv(TRAIN_PATH, low_memory=False)

if TARGET_COL not in df.columns:
    raise ValueError("TARGET column missing.")

y_series = pd.to_numeric(df[TARGET_COL], errors="coerce")

if y_series.isna().any():
    raise ValueError("TARGET contains missing or non-numeric values.")

if sorted(y_series.unique().tolist()) != [0, 1]:
    raise ValueError("Expected binary TARGET values [0, 1].")

y = y_series.astype("int8").to_numpy()

# ------------------------------------------------------------
# 2. Reproduce original baseline feature preparation
# ------------------------------------------------------------
work_df = df.drop(columns=[TARGET_COL])

dropped_cols = [
    c for c in EXCLUDE_COLUMNS
    if c in work_df.columns
]

feature_cols = [
    c for c in work_df.columns
    if c not in dropped_cols
]

X = work_df[feature_cols].copy()

categorical_cols = []
numeric_cols = []
converted_cols = []

for col in X.columns:
    s = X[col]

    if (
        pd.api.types.is_object_dtype(s.dtype)
        or isinstance(s.dtype, pd.CategoricalDtype)
        or pd.api.types.is_string_dtype(s.dtype)
    ):
        X[col] = (
            s.astype("object")
             .where(s.notna(), "__MISSING__")
             .astype(str)
        )
        categorical_cols.append(col)

    elif pd.api.types.is_bool_dtype(s.dtype):
        X[col] = s.astype("int8")
        numeric_cols.append(col)

    elif pd.api.types.is_numeric_dtype(s.dtype):
        X[col] = pd.to_numeric(
            s, errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        numeric_cols.append(col)

    else:
        X[col] = (
            s.astype("object")
             .where(s.notna(), "__MISSING__")
             .astype(str)
        )
        categorical_cols.append(col)
        converted_cols.append(col)

# Original baseline constant-feature removal
constant_cols = [
    c for c in X.columns
    if X[c].nunique(dropna=False) <= 1
]

if constant_cols:
    X = X.drop(columns=constant_cols)
    categorical_cols = [
        c for c in categorical_cols
        if c not in constant_cols
    ]
    numeric_cols = [
        c for c in numeric_cols
        if c not in constant_cols
    ]
    dropped_cols.extend(constant_cols)

dropped_cols.extend(converted_cols)

if X.isna().all(axis=0).any():
    raise ValueError("An entirely missing predictor was found.")

if len(X) != len(y):
    raise ValueError("X/y row mismatch.")

if len(X) != 171202 or X.shape[1] != 31:
    raise ValueError(
        f"Unexpected dataset/schema: X={X.shape}, y={len(y)}"
    )

print("DATA VALIDATED")
print("Rows:", len(X))
print("Predictors:", X.shape[1])
print("Categorical:", len(categorical_cols))
print("Numeric:", len(numeric_cols))
print("Positive rate:", y.mean())

# ------------------------------------------------------------
# 3. Recreate exact repeated stratified fold assignments
#    No CatBoost training is performed here.
# ------------------------------------------------------------
cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=BASE_SEED
)

fold_assignments = np.full(
    (len(X), N_REPEATS),
    -1,
    dtype=np.int8
)

for split_number, (_, valid_idx) in enumerate(cv.split(X, y)):
    repetition = split_number // N_SPLITS
    fold = split_number % N_SPLITS + 1

    fold_assignments[valid_idx, repetition] = fold

if np.any(fold_assignments < 0):
    raise RuntimeError("Some rows did not receive fold assignments.")

# ------------------------------------------------------------
# 4. Verify Repetition 1 against the five existing Stage II-B ZIPs
# ------------------------------------------------------------
id_values = df[ID_COL].to_numpy()

for fold in range(1, 6):

    zip_path = ZIP_DIR / f"Stage_II_B_31_r1_f{fold}.zip"

    if not zip_path.is_file():
        raise FileNotFoundError(
            f"Cannot verify missing archive: {zip_path}"
        )

    expected_idx = np.flatnonzero(
        fold_assignments[:, 0] == fold
    )

    with zipfile.ZipFile(zip_path) as z:

        matches = [
            m for m in z.namelist()
            if m.endswith("Dependency_Validation.csv.gz")
        ]

        if len(matches) != 1:
            raise RuntimeError(
                f"{zip_path.name}: expected one validation CSV, "
                f"found {matches}"
            )

        observed = pd.read_csv(
            z.open(matches[0]),
            compression="gzip"
        )

    required = {"row_index", ID_COL}

    if not required.issubset(observed.columns):
        raise RuntimeError(
            f"{zip_path.name}: missing row_index or applicant ID."
        )

    observed["row_index"] = observed["row_index"].astype("int64")

    observed = (
        observed[["row_index", ID_COL]]
        .sort_values("row_index")
        .reset_index(drop=True)
    )

    expected = pd.DataFrame({
        "row_index": expected_idx,
        ID_COL: id_values[expected_idx]
    })

    try:
        pd.testing.assert_frame_equal(
            observed,
            expected,
            check_dtype=False,
            check_exact=True
        )
    except AssertionError as exc:
        raise RuntimeError(
            f"FOLD VERIFICATION FAILED: repetition 1, fold {fold}.\n"
            "Do not use the regenerated assignments.\n"
            f"{exc}"
        ) from exc

    print(
        f"r1_f{fold}: EXACT MATCH | "
        f"{len(expected_idx):,} validation applicants"
    )

print("\nALL FIVE REPETITION 1 FOLDS VERIFIED.")

# ------------------------------------------------------------
# 5. Rebuild original baseline artifact files
# ------------------------------------------------------------
assignments = pd.DataFrame({
    "row_index": np.arange(len(df)),
    ID_COL: id_values,
    "TARGET": y
})

for r in range(N_REPEATS):
    assignments[f"repeat_{r + 1}_fold"] = (
        fold_assignments[:, r]
    )

config = {
    "input_file": str(TRAIN_PATH),
    "rows": int(len(X)),
    "target": TARGET_COL,
    "excluded_columns": dropped_cols,
    "predictor_columns": X.columns.tolist(),
    "categorical_columns": categorical_cols,
    "numeric_columns": numeric_cols,
    "positive_rate": float(y.mean()),
    "cv": {
        "type": "RepeatedStratifiedKFold",
        "n_splits": N_SPLITS,
        "n_repeats": N_REPEATS,
        "random_state": BASE_SEED
    },
    "catboost": {
        "iterations": 1000,
        "depth": 6,
        "learning_rate": 0.05,
        "l2_leaf_reg": 3.0,
        "loss_function": "Logloss",
        "eval_metric": "AUC",
        "thread_count": 4,
        "early_stopping": False
    },
    "note": (
        "Baseline with original predictors only. Target is not used "
        "in preprocessing or feature construction. Outer validation "
        "folds are not used for early stopping."
    )
}

config_path = DEST / "Baseline_Run_Config.json"
fold_path = DEST / "Baseline_CV_Fold_Assignments.csv"

with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

assignments.to_csv(fold_path, index=False)

print("\nRECOVERED ARTIFACTS:")
print(config_path)
print(fold_path)

print("\nVALIDATION:")
print("Rows:", len(assignments))
print("Predictors:", len(config["predictor_columns"]))
print("Repetitions:", N_REPEATS)
print("Folds per repetition:", N_SPLITS)
print("TARGET in predictor list:", TARGET_COL in X.columns)
print("All fold labels valid:",
      set(np.unique(fold_assignments.ravel())) == {1, 2, 3, 4, 5})

print("\nYou can now resume Stage II-B.")

```

```python

import json
import zipfile
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.model_selection import RepeatedStratifiedKFold

# ------------------------------------------------------------
# Fixed values taken from the original baseline script
# ------------------------------------------------------------
BASE_SEED = 20261031
N_SPLITS = 5
N_REPEATS = 3

TARGET_COL = "TARGET"
ID_COL = "SK_ID_CURR"
EXCLUDE_COLUMNS = [ID_COL, "FLAG_MOBIL"]

TRAIN_PATH = Path(
    "/content/drive/MyDrive/Competition/input/train.csv"
)

ZIP_DIR = Path("/content/drive/MyDrive/Competition")

DEST = Path("/content/baseline_artifacts")
DEST.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Load the original dataset
# ------------------------------------------------------------
if not TRAIN_PATH.is_file():
    raise FileNotFoundError(f"Missing input: {TRAIN_PATH}")

df = pd.read_csv(TRAIN_PATH, low_memory=False)

if TARGET_COL not in df.columns:
    raise ValueError("TARGET column missing.")

y_series = pd.to_numeric(df[TARGET_COL], errors="coerce")

if y_series.isna().any():
    raise ValueError("TARGET contains missing or non-numeric values.")

if sorted(y_series.unique().tolist()) != [0, 1]:
    raise ValueError("Expected binary TARGET values [0, 1].")

y = y_series.astype("int8").to_numpy()

# ------------------------------------------------------------
# 2. Reproduce original baseline feature preparation
# ------------------------------------------------------------
work_df = df.drop(columns=[TARGET_COL])

dropped_cols = [
    c for c in EXCLUDE_COLUMNS
    if c in work_df.columns
]

feature_cols = [
    c for c in work_df.columns
    if c not in dropped_cols
]

X = work_df[feature_cols].copy()

categorical_cols = []
numeric_cols = []
converted_cols = []

for col in X.columns:
    s = X[col]

    if (
        pd.api.types.is_object_dtype(s.dtype)
        or isinstance(s.dtype, pd.CategoricalDtype)
        or pd.api.types.is_string_dtype(s.dtype)
    ):
        X[col] = (
            s.astype("object")
             .where(s.notna(), "__MISSING__")
             .astype(str)
        )
        categorical_cols.append(col)

    elif pd.api.types.is_bool_dtype(s.dtype):
        X[col] = s.astype("int8")
        numeric_cols.append(col)

    elif pd.api.types.is_numeric_dtype(s.dtype):
        X[col] = pd.to_numeric(
            s, errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        numeric_cols.append(col)

    else:
        X[col] = (
            s.astype("object")
             .where(s.notna(), "__MISSING__")
             .astype(str)
        )
        categorical_cols.append(col)
        converted_cols.append(col)

# Original baseline constant-feature removal
constant_cols = [
    c for c in X.columns
    if X[c].nunique(dropna=False) <= 1
]

if constant_cols:
    X = X.drop(columns=constant_cols)
    categorical_cols = [
        c for c in categorical_cols
        if c not in constant_cols
    ]
    numeric_cols = [
        c for c in numeric_cols
        if c not in constant_cols
    ]
    dropped_cols.extend(constant_cols)

dropped_cols.extend(converted_cols)

if X.isna().all(axis=0).any():
    raise ValueError("An entirely missing predictor was found.")

if len(X) != len(y):
    raise ValueError("X/y row mismatch.")

if len(X) != 171202 or X.shape[1] != 31:
    raise ValueError(
        f"Unexpected dataset/schema: X={X.shape}, y={len(y)}"
    )

print("DATA VALIDATED")
print("Rows:", len(X))
print("Predictors:", X.shape[1])
print("Categorical:", len(categorical_cols))
print("Numeric:", len(numeric_cols))
print("Positive rate:", y.mean())

# ------------------------------------------------------------
# 3. Recreate exact repeated stratified fold assignments
#    No CatBoost training is performed here.
# ------------------------------------------------------------
cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=BASE_SEED
)

fold_assignments = np.full(
    (len(X), N_REPEATS),
    -1,
    dtype=np.int8
)

for split_number, (_, valid_idx) in enumerate(cv.split(X, y)):
    repetition = split_number // N_SPLITS
    fold = split_number % N_SPLITS + 1

    fold_assignments[valid_idx, repetition] = fold

if np.any(fold_assignments < 0):
    raise RuntimeError("Some rows did not receive fold assignments.")

# ------------------------------------------------------------
# 4. Verify Repetition 1 against the five existing Stage II-B ZIPs
# ------------------------------------------------------------
id_values = df[ID_COL].to_numpy()

for fold in range(1, 6):

    zip_path = ZIP_DIR / f"Stage_II_B_31_r1_f{fold}.zip"

    if not zip_path.is_file():
        raise FileNotFoundError(
            f"Cannot verify missing archive: {zip_path}"
        )

    expected_idx = np.flatnonzero(
        fold_assignments[:, 0] == fold
    )

    with zipfile.ZipFile(zip_path) as z:

        matches = [
            m for m in z.namelist()
            if m.endswith("Dependency_Validation.csv.gz")
        ]

        if len(matches) != 1:
            raise RuntimeError(
                f"{zip_path.name}: expected one validation CSV, "
                f"found {matches}"
            )

        observed = pd.read_csv(
            z.open(matches[0]),
            compression="gzip"
        )

    required = {"row_index", ID_COL}

    if not required.issubset(observed.columns):
        raise RuntimeError(
            f"{zip_path.name}: missing row_index or applicant ID."
        )

    observed["row_index"] = observed["row_index"].astype("int64")

    observed = (
        observed[["row_index", ID_COL]]
        .sort_values("row_index")
        .reset_index(drop=True)
    )

    expected = pd.DataFrame({
        "row_index": expected_idx,
        ID_COL: id_values[expected_idx]
    })

    try:
        pd.testing.assert_frame_equal(
            observed,
            expected,
            check_dtype=False,
            check_exact=True
        )
    except AssertionError as exc:
        raise RuntimeError(
            f"FOLD VERIFICATION FAILED: repetition 1, fold {fold}.\n"
            "Do not use the regenerated assignments.\n"
            f"{exc}"
        ) from exc

    print(
        f"r1_f{fold}: EXACT MATCH | "
        f"{len(expected_idx):,} validation applicants"
    )

print("\nALL FIVE REPETITION 1 FOLDS VERIFIED.")

# ------------------------------------------------------------
# 5. Rebuild original baseline artifact files
# ------------------------------------------------------------
assignments = pd.DataFrame({
    "row_index": np.arange(len(df)),
    ID_COL: id_values,
    "TARGET": y
})

for r in range(N_REPEATS):
    assignments[f"repeat_{r + 1}_fold"] = (
        fold_assignments[:, r]
    )

config = {
    "input_file": str(TRAIN_PATH),
    "rows": int(len(X)),
    "target": TARGET_COL,
    "excluded_columns": dropped_cols,
    "predictor_columns": X.columns.tolist(),
    "categorical_columns": categorical_cols,
    "numeric_columns": numeric_cols,
    "positive_rate": float(y.mean()),
    "cv": {
        "type": "RepeatedStratifiedKFold",
        "n_splits": N_SPLITS,
        "n_repeats": N_REPEATS,
        "random_state": BASE_SEED
    },
    "catboost": {
        "iterations": 1000,
        "depth": 6,
        "learning_rate": 0.05,
        "l2_leaf_reg": 3.0,
        "loss_function": "Logloss",
        "eval_metric": "AUC",
        "thread_count": 4,
        "early_stopping": False
    },
    "note": (
        "Baseline with original predictors only. Target is not used "
        "in preprocessing or feature construction. Outer validation "
        "folds are not used for early stopping."
    )
}

config_path = DEST / "Baseline_Run_Config.json"
fold_path = DEST / "Baseline_CV_Fold_Assignments.csv"

with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

assignments.to_csv(fold_path, index=False)

print("\nRECOVERED ARTIFACTS:")
print(config_path)
print(fold_path)

print("\nVALIDATION:")
print("Rows:", len(assignments))
print("Predictors:", len(config["predictor_columns"]))
print("Repetitions:", N_REPEATS)
print("Folds per repetition:", N_SPLITS)
print("TARGET in predictor list:", TARGET_COL in X.columns)
print("All fold labels valid:",
      set(np.unique(fold_assignments.ravel())) == {1, 2, 3, 4, 5})

print("\nYou can now resume Stage II-B.")

```

```python
import zipfile
import numpy as np
import pandas as pd
from pathlib import Path

FN = "Baseline_CV_Fold_Assignments.csv"
fold_file = next((p for p in [
    Path("/content/baseline_artifacts") / FN,
    Path("/content/drive/MyDrive/Competition") / FN,
] if p.is_file()), None)
if fold_file is None:
    raise FileNotFoundError("Run the recovery script first.")

folds = pd.read_csv(fold_file)
ids = folds["SK_ID_CURR"].to_numpy()
n = len(folds)
print("Using:", fold_file, "| rows:", n)

search_dirs = [Path("/content"), Path("/content/drive/MyDrive/Competition")]
results = []

for r in (1, 2, 3):
    col = f"repeat_{r}_fold"
    for f in range(1, 6):
        name = f"Stage_II_B_31_r{r}_f{f}.zip"
        zp = next((d / name for d in search_dirs if (d / name).is_file()), None)
        if zp is None:
            results.append((r, f, "ZIP NOT FOUND", "", "")); continue

        exp_valid = np.flatnonzero(folds[col].to_numpy() == f)
        exp_train = np.flatnonzero(folds[col].to_numpy() != f)

        with zipfile.ZipFile(zp) as z:
            def get(suffix):
                m = [x for x in z.namelist() if x.endswith(suffix)]
                if len(m) != 1:
                    raise RuntimeError(f"{name}: {suffix} -> {m}")
                return pd.read_csv(z.open(m[0]), compression="gzip")
            valid = get("Dependency_Validation.csv.gz")
            train = get("Dependency_Train_OOF.csv.gz")

        v_idx = np.sort(valid["row_index"].to_numpy())
        t_idx = np.sort(train["row_index"].to_numpy())
        v_ok = np.array_equal(v_idx, exp_valid)
        t_ok = np.array_equal(t_idx, exp_train)
        id_ok = np.array_equal(
            valid.sort_values("row_index")["SK_ID_CURR"].to_numpy(),
            ids[exp_valid])
        results.append((r, f, "EXACT MATCH" if (v_ok and t_ok and id_ok) else "MISMATCH",
                        f"valid={len(v_idx):,}", f"train={len(t_idx):,}"))

out = pd.DataFrame(results, columns=["rep", "fold", "status", "valid_rows", "train_rows"])
print(out.to_string(index=False))
print("\nALL 15 VERIFIED:", (out["status"] == "EXACT MATCH").all())
```
